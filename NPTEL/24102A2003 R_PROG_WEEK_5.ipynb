## **Find the positions of missing values**

```r
x <- c(11,NA,13,NA) #combine or concatenate

which(is.na(x))
```

```r
sum(is.na(x))
```

```r
complete.cases(x) # It checks each position to see if it contains valid, non-missing data.
```

```r
y <- na.omit(x) # remove missing values

y
```

```r
attr(y, "na.action")
```

```r
# Compare mean before and after
mean(x)
mean(y)
```

## **Nested if...else if...else**

```r
x <- 5

if (x == 3) {
    x <- x - 1
} else if (x < 3) {
    x <- x + 5
} else {
    x <- 2 * x
}

x
```

```r
x <- 3

if (x == 3) {
    x <- x - 1
} else if (x < 3) {
    x <- x + 5
} else {
    x <- 2 * x
}

x
```

## **ifelse with vector**

```r
x <- 1:10

ifelse(x < 6, x^2, x + 1)
```

```r
x <- c(7, 9, 8, 4)

ifelse(x %% 2 == 0, "even number", "odd number")
```
